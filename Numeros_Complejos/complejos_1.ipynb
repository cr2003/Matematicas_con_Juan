### Video 2: Pasar a forma Polar


```python
# Example
# modulus (r), argument (theta)
# Define the imaginary unit and complex number
i = sp.I 
z = 1 + i * sp.sqrt(3)  # This is a classic 1, sqrt(3), 2 triangle

# 1. Get the components
r = sp.Abs(z)
theta = sp.arg(z)

print(f"Modulus (r): {r}")
print(f"Argument (theta): {theta}")

# 2. Represent in Polar Form (r * e^(i*theta))
polar_form = r * sp.exp(i * theta)
print(f"Polar Form: {polar_form}")
polar_form
```

### Video 9: Hallar las Raíces

```python
# z**3 + 1 = 0
z = sp.symbols("z")
eq = sp.Eq(z**3 + 1, 0)
sp.solve(eq, z)
```

### Video 10: Diferentes representaciones de un número complejo

```python
i = sp.I
e = sp.E
pi = sp.pi
```

#### Forma exponencial:

```python
z = 5*e**(i*pi/6)
z
```

#### Formula de Euler

$$e^{i \theta} = \cos \theta + i \sin \theta$$

We can convert the exponential form to binomial form

In our previous example:

$$z = 5 e^{i \frac{\pi}{6}} = 5 \left[\cos \frac{\pi}{6} + i \sin \frac{\pi}{6}\right] $$

```python
# Using SymPy
# Function to obtain the modulus
r = sp.Abs(z)
print(f"Modulus z: {r}")
# Function to obtain the argument
theta = sp.arg(z)
print(f"Argument z: {theta}")

z_binomial = r * (sp.cos(theta) + i * sp.sin(theta))
z_binomial
```

```python
# In SymPy there is a way to convet directly
z_binomial_2 = z.rewrite(sp.cos)
z_binomial_2
```

```python
# Polar form
print(f"Polar form: ({sp.Abs(z)},{sp.arg(z)})")
```

```python
# The expoential form
r*sp.exp(i*theta)
```
