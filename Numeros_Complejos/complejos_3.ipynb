#### Distintas formas de escribir un número complejo
##### Forma binómica:
$$z = a + bi$$

##### Forma cartesiana:
$$z = (a, b)$$

##### Forma polar:
**Módulo**
$$|z| = \sqrt{a^2 + b^2}$$

**Argumento**
$$arg = \theta + 2\pi k $$

$$z = |z|_{\theta + 2\pi k}$$

##### Forma trigonométrica:

$$z = |z|(\cos \theta + i \sin \theta)$$

##### Forma exponencial:

**Ecuación de Euler**
$$e^{i\theta} = \cos \theta + i \sin \theta$$

Substituyendo $\cos \theta + i \sin \theta$ en la forma trigonométrica, 
nos queda la siguiente forma exponencial:

$$z = |z| e^{i\theta}$$

#### Ejercicio:
Partiendo de un número complejo z, en forma binómica, lo vamos a expresar en las distintas formas.

```python
# Forma binómica
z = 1 + i
z
```

```python
# Forma cartesiana
print(f"En forma cartesiana z=({sp.re(z)}, {sp.im(z)}) ")
```

```python
# Forma polar
# modulo
r = sp.Abs(z)
# argumento
theta = sp.arg(z)
print(f"En forma polar z=({r}, {theta}) ")
```

```python
# Forma trigonométrica
z_trig = r * (sp.cos(theta)+i*sp.sin(theta))
z_trig
```

```python
# Forma exponencial
z_exp=r*sp.exp(i*theta)
z_exp
```
