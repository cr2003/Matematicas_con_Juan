### Video 22: Determinar el valor de a  


```python
# Define the symbols
# we assume 'a' is a real number for the context of the problem
a = sp.symbols('a', real=True)

# Define the expression
expr = (2 + a*i) * (1 + i)

# Expand the expression
expanded_expr = sp.expand(expr)

# Extract the imaginary part
imag_part = sp.im(expanded_expr)

# Solve for 'a' where the imaginary part is 0
solution = sp.solve(imag_part, a)

print(f"Expanded expression: {expanded_expr}")
print(f"Imaginary part: {imag_part}")
print(f"The value of 'a' for the result to be real is: {solution[0]}")
```
