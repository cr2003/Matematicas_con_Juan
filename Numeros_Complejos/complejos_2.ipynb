### Video 11: Pasar the forma binómica a trigonometrica

Forma binómica:
$$z = a + bi$$

Forma trigonométrica:
$$z = |z|\cos\theta + i |z|\sin\theta$$

$$z = |z| (\cos\theta + i \sin\theta)$$


```python
z = 1 + sp.sqrt(3)* i
z
```

```python
# modulus
r = sp.Abs(z)
print(f"Modulus z = {r}")

# Argument
theta = sp.arg(z)
theta
```

```python
z_trig = r * sp.cos(theta) + i * r * sp.sin(theta)
print(f"Forma trigonometrica z = {r} * (cos {theta} + i sin {theta})")
```

### Video 14: Formula de Euler

$$e^{i \theta} = \cos \theta + i \sin \theta$$

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup data
theta = np.linspace(0, 4 * np.pi, 200)  # Two full rotations
real_part = np.cos(theta)
imag_part = np.sin(theta)

# 2. Create 3D plot
# fig = plt.figure(figsize=(10, 7))
fig = plt.figure(figsize=(20, 9))

ax = fig.add_subplot(111, projection='3d')

# Plot the 3D Helix (e^îθ)
ax.plot(theta, real_part, imag_part, label=r'$e^{i\theta}$', lw=2, color='blue')

# Plot the projections (Shadows)
ax.plot(theta, real_part, -1.5, label='Real projection (Cosine)', color='red', alpha=0.5)
ax.plot(theta, imag_part, 1.5, label='Imaginary projection (Sine)', color='green', alpha=0.5)

# Labels and formatting
ax.set_xlabel('Theta (θ)')
ax.set_ylabel('Real (cos θ)')
ax.set_zlabel('Imaginary (sin θ)')
ax.set_title("Euler's Formula: The Complex Helix")
ax.set_ylim(-1.5, 1.5)
ax.set_zlim(-1.5, 1.5)
ax.legend()
fig.subplots_adjust(left=0, right=1, bottom=0, top=1)
plt.tight_layout()
plt.show()
```
