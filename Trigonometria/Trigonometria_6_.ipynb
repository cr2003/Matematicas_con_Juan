```python
def bearing_math_angle(angle):

    """
    Converts a navigational bearing (0=North, clockwise)
    to an angle (0=East, counter-clockwise) (math angle)
    and viceversa
    """
    # Normalize bearing to 0-360 range
    angle %= 360

    # Apply conversion
    conversion_angle = (90 - angle) % 360

    return conversion_angle
```

```python
def fmt_vec(v, precision=2, type="f"):
    """Format the output of a vector with the specified precision"""
    return (
        f"<{v.x:.{precision}{type}}, {v.y:.{precision}{type}}, {v.z:.{precision}{type}}>"
    )
```

### Hallar la altura del árbol

```python
import sympy as sp

# Define our variable
h,x = sp.symbols("h,x")

# Define angles from the Horizontal in radians
# dog angle = 40
# other angle =  30 
angle_dog = sp.rad(40)
angle_other = sp.rad(30)

# x distance from dog to the tree, h heigh of the tree
# Based on tan(theta) = h / x
# h = x * tan(40)
# h = (x + 5) * tan(30)
eq1 = sp.Eq(h, x * sp.tan(angle_dog))
eq1
```

```python
eq2 = sp.Eq(h,  (x + 5) * sp.tan(angle_other))
eq2
```

```python
# Solving for h, x
solution = sp.solve([eq1, eq2],[x,h], dict=True)
solution
```

```python
print(f"Numerical result for h={float(solution[0][h]):.2f} meters")
print(f"Numerical result for x={float(solution[0][x]):.2f} meters")
```
