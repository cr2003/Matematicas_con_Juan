### HALLAR LA ALTURA DEL PERRO

```python
import sympy as sp
from IPython.display import display, Math

# 1. Define symbols
h = sp.Symbol('h', positive=True)
hyp = 50

# 2. Define angles in radians
angle = sp.rad(37)


# 3. Set up the equation
equation = sp.Eq(h, hyp * sp.sin(angle))

# 4. Solve for h
h_sol = sp.solve(equation, h)[0]

# 5. Display results
display(Math(rf"h \approx {h_sol.evalf(2)} \text{{ cm}}"))
```
