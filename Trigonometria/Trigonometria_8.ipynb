### HALLAR LA HIPOTENUSA Y EL CATETO. 

<img src="../pics/Trig_video_8.png" width=600> 


```python
import sympy as sp
from IPython.display import display, Math

# 1. Define symbols
h = sp.Symbol('h', positive=True)
total_dist = 100

# 2. Define angles in radians
angle_left = sp.rad(30)
angle_right = sp.rad(50)

# 3. Set up the equation
equation = sp.Eq(total_dist, h/sp.tan(angle_left) + h/sp.tan(angle_right))

# 4. Solve for h
h_sol = sp.solve(equation, h)[0]

# 5. Display results
display(Math(rf"h \approx {h_sol.evalf(4)} \text{{ m}}"))
```
