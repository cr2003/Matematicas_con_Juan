```python
import sympy as sp
from IPython.display import display, Math

# Inicializar la impresión elegante de SymPy
sp.init_printing(use_latex='mathjax')

# Definir símbolos
x, b, c = sp.symbols('x b c')
a = 1  # Valor dado

# 1. Resolver x (Suma de ángulos: 90 + 2x + x = 180)
ecuacion_x = sp.Eq(90 + 3*x, 180)
solucion_x = sp.solve(ecuacion_x, x)[0]

display(Math(rf"\text{{1. Valor de }} x: x = {solucion_x}^\circ"))

# Convertir x a radianes para cálculos trigonométricos
x_rad = sp.rad(solucion_x)

# 2. Resolver b (Usando tan(x) = a / b)
# Definimos la ecuación: tan(30°) = 1 / b
ecuacion_b = sp.Eq(sp.tan(x_rad), a / b)
solucion_b = sp.solve(ecuacion_b, b)[0]

display(Math(rf"\text{{2. Lado }} b: b = {sp.latex(solucion_b)} \approx {solucion_b.evalf(4)} \text{{ cm}}"))

# 3. Resolver c (Usando sin(x) = a / c)
# Definimos la ecuación: sin(30°) = 1 / c
ecuacion_c = sp.Eq(sp.sin(x_rad), a / c)
solucion_c = sp.solve(ecuacion_c, c)[0]

display(Math(rf"\text{{3. Lado }} c: c = {sp.latex(solucion_c)} \text{{ cm}}"))

```
