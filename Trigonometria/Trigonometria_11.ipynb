```python
import sympy as sp
from IPython.display import display, Math

# 1. Define symbols and the square diagonal
d = sp.Integer(8)
angle = sp.rad(30)
X = sp.Symbol('X', positive=True)

# 2. Calculate the side of the square (the 'red side')
red_side = d / sp.sqrt(2)

# 3. Set up the triangle equation: sin(30) = red_side / X
equation = sp.Eq(sp.sin(angle), red_side / X)

# 4. Solve for the hypotenuse X
X_sol = sp.solve(equation, X)[0]

# 5. Display results
display(Math(rf"\text{{Side of square}} = {sp.latex(red_side.simplify())}\text{{ cm}}"))
display(Math(rf"X = {sp.latex(X_sol.simplify())} \approx {X_sol.evalf(4)}\text{{ cm}}"))
```
