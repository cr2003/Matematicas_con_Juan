### HALLAR LA HIPOTENUSA Y EL CATETO. 

<img src="../pics/Trig_video_7.png" width=400> 


🦉 **Note:** In the video the side a = 2 cm.

```python
import sympy as sp
from IPython.display import display, Math

# 1. Define the angle and the known side
# We use rad() because SymPy trig functions expect radians
angle = sp.rad(30) 
a = sp.Integer(2)
b, c = sp.symbols('b c', positive=True)

# 2. Set up the equations based on Trig Ratios
# sin(30) = a/c  =>  c = a/sin(30)
# tan(30) = a/b  =>  b = a/tan(30)
eqn_c = sp.Eq(sp.sin(angle), a/c)
eqn_b = sp.Eq(sp.tan(angle), a/b)

# 3. Solve the equations
sol_c = sp.solve(eqn_c, c)[0]
sol_b = sp.solve(eqn_b, b)[0]

# 4. Display the exact and numerical results
display(Math(rf"c = {sp.latex(sol_c)} = {sol_c.evalf(3)}\text{{ cm}}"))
display(Math(rf"b = {sp.latex(sol_b)} = {sol_b.evalf(3)}\text{{ cm}}"))
```
