```python
from vpython import *
from astropy import units as u
```

```python
def bearing_math_angle(angle):

    """
    Converts a navigational bearing (0=North, clockwise)
    to an angle (0=East, counter-clockwise) (math angle)
    and viceversa
    """
    # Normalize bearing to 0-360 range
    angle %= 360

    # Apply conversion
    conversion_angle = (90 - angle) % 360

    return conversion_angle
```

```python
def fmt_vec(v, precision=2, type="f"):
    """Format the output of a vector with the specified precision"""
    return (
        f"<{v.x:.{precision}{type}}, {v.y:.{precision}{type}}, {v.z:.{precision}{type}}>"
    )
```

## Video 1
### Hallar la distancia entre dos puntos dado el rumbo

Conducimos 42 km con rumbo $32^\circ$, después cambiamos al rumbo $154^\circ$ y conducimos otros 28 km. Queremos calcular la distancia que hemos recorrido.

```python
bearing_1 = 32      # deg
r_1_mag = 42        # km
bearing_2 = 154     # deg
r_2_mag = 28         # km

theta_1_math = bearing_math_angle(bearing_1)
theta_1_math_rad = radians(theta_1_math)
r1 = vec(r_1_mag * cos(theta_1_math_rad), r_1_mag * sin(theta_1_math_rad), 0)


theta_2_math = bearing_math_angle(bearing_2)
theta_2_math_rad = radians(theta_2_math)
r2 = vec(r_2_mag * cos(theta_2_math_rad), r_2_mag * sin(theta_2_math_rad), 0)

r = r1 + r2
r_mag = r.mag
theta_rad = atan2(r.y, r.x)
theta_deg = degrees(theta_rad)
bearing = bearing_math_angle(theta_deg)

print(f"El vector de distacia es r={fmt_vec(r,precision=1)} km")
print(f"La distacia es r_mag={r.mag:.1f} km")
print(f"El  rumbo bearing={bearing:.1f} deg")
 
```

### Hallar el valor de la altura
<img src="../pics/Trig_video_2.png" width=600> 


```python
# En el triangulo inferior, calculamos el angulo theta
theta_rad = atan2(8,30)
theta_deg = degrees(theta_rad)

# Calculamos el angulo total como la suma the theta y 22
ang_total_deg = 22 + theta_deg

# Calculamos la altura del triangulo total h; tan(ang_total) = h / 30 ==> h = 30 * tan(ang_total)
ang_total_rad = radians(ang_total_deg)
h = 30 * tan(ang_total_rad)

# Con esta altura calculamos y; h = y + 8 ==> y = h - 8
y = h - 8
print(f"y = {y:.2f} m")
```

#### Tarea

<img src="../pics/Video_2_tarea.png" width=600> 



```python
import sympy as sp

# Define our variable
x = sp.symbols("x")

# Define angles from the VERTICAL  in radians
# Bottom angle (to the lower point) = 43
# Top angle (to the higher point) =  30 
angle_low = sp.rad(43)
angle_hig = sp.rad(30)

# d distance to the building, h heigh of the building
# Based on tan(theta) = d / h
# d = x * tan(43)
# d = (x + 5) * tan(73)
equation = sp.Eq(x * sp.tan(angle_low), (x + 5) * sp.tan(angle_hig))
equation
```

```python
# Solving for x
solution = sp.solve(equation, x)[0]
solution
```

```python
print(f"Exact expression for x {solution}")
print(f"Numerical result {float(solution):.2f} meters")
```

#### 1.- grados $\implies$ radianes

a) $45^\circ \implies 45^\circ \frac{\pi}{180^\circ} = \frac{\cancel{45^\circ } \pi}{4 \cancel{\times 45^\circ}} = \frac{\pi}{4}$

b) $60^\circ \implies 60^\circ \frac{\pi}{180^\circ} = \frac{\cancel{60^\circ } \pi}{3 \cancel{\times 60^\circ}} = \frac{\pi}{3}$



#### 2.- radianes $\implies$ grados

a) $\frac{\pi}{6} \implies \frac{\pi}{6} \times \frac{180^\circ}{\pi} = \frac{\cancel{\pi \;6} \times 30^\circ}{\cancel{6 \pi}} = 30^\circ$
 


b) $\frac{5\pi}{4} \implies \frac{5 \cancel{\pi}}{4} \times \frac{180^\circ}{\cancel{\pi}} = \frac{5 \times \cancel{4} \times 45^\circ}{\cancel{4}} = 225^\circ $
