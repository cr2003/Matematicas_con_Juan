```python
import sympy as sp
from IPython.display import display, Math


# Definir la variable simbólica
h = sp.symbols('h', positive=True)

# Definir ángulos en radianes
tan_15 = sp.tan(sp.rad(15))
tan_30 = sp.tan(sp.rad(30))

# Expresión de distancias
d1 = h / tan_15  # para 15°
d2 = h / tan_30  # para 30°

# Ecuación: diferencia de distancias = 20
equation = sp.Eq(d1 - d2, 20)

# Resolver la ecuación
solution = sp.solve(equation, h)[0]

# Mostrar resultado
display(Math(rf"\text{{h}} = {sp.latex(solution)}\text{{ m}}"))
```
