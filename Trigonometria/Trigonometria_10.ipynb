```python
import sympy as sp
from IPython.display import display, Math

# 1. Define symbols
h = sp.Symbol('h', positive=True)

# 2. Define angles in radians
angle = sp.rad(30)


# 3. Set up the equation
equation = sp.Eq(sp.cos(angle), 10 / h )

# 4. Solve for h
h_sol = sp.solve(equation, h)[0]

# 5. Display results
display(h_sol)
display(Math(rf"h \approx {h_sol.evalf(4)} \text{{ cm}}"))
```

```python
import sympy as sp
from IPython.display import display, Math

# 1. Define symbols
theta = sp.Symbol('theta', positive=True)

# 3. Set up the equation
equation = sp.Eq(sp.sin(theta), 5 / 10)

# 4. Solve for theta
theta_sol = sp.solve(equation, theta)[0]
theta_sol_deg = degrees(theta_sol)

# 5. Display results
display(Math(rf"theta \approx {theta_sol.evalf(4)} \text{{ rad}}"))
display(Math(rf"theta = {theta_sol_deg:.0f} \text{{°}}"))


```

```python
import sympy as sp
from IPython.display import display, Math

# 1. Define symbols
theta = sp.Symbol('theta', positive=True)

# 3. Set up the equation
equation = sp.Eq(sp.cos(theta), 7 / 14)

# 4. Solve for theta
theta_sol = sp.solve(equation, theta)[0]
theta_sol_deg = degrees(theta_sol)

# 5. Display results
display(Math(rf"theta \approx {theta_sol.evalf(4)} \text{{ rad}}"))
display(Math(rf"theta = {theta_sol_deg:.0f} \text{{°}}"))
```
