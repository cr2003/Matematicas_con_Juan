# Video 16 - Simplificar
## 1)
$$\frac{2x + 2y}{x + y}$$

```python
expr_raw = r"\frac{2x + 2y}{x + y}"
expr_sp =(2*x + 2*y) / (x + y)
sol = expr_sp.simplify()
print("raw expr:   \tSymPy expr:  \tsolution:")
print("-" * 40)
# display(Math(expr_raw + "=" + sp.latex(expr_sp) + "=\t" + sp.latex(sol)))

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad" # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))


```

## 2)
$$\frac{2x + y}{x +\frac{1}{2}y}$$

```python
expr_raw = r"\frac{2x + y}{x +\frac{1}{2}y}"
expr_sp =(2*x + y) / (x + sp.Rational(1,2) * y)
sol = expr_sp.simplify()
print("raw expr:   \tSymPy expr:  \tsolution:")
print("-" * 40)
# display(Math(expr_raw + "=" + sp.latex(expr_sp) + "=\t" + sp.latex(sol)))

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad" # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))


```

## 3)
$$\frac{\frac{3}{5}x + 2y}{x +\frac{10}{3}y}$$

```python
expr_raw = r"\frac{\frac{3}{5}x + 2y}{x +\frac{10}{3}y}"
expr_sp =(sp.Rational(3,5)*x + 2*y) / (x + sp.Rational(10,3) * y)
sol = expr_sp.simplify()
print("raw expr:   \tSymPy expr:  \tsolution:")
print("-" * 40)
# display(Math(expr_raw + "=" + sp.latex(expr_sp) + "=\t" + sp.latex(sol)))

# Nice display
spacing = r"\quad = \qquad" # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))


```

## 4)
$$\frac{7x + y}{\frac{7}{5} +\frac{y}{5x}}$$
