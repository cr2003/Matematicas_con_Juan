```python
from vpython import *
from astropy import units as u
import sympy as sp
from IPython.display import display, Math

a, m, p, q, r, s, t, u, v, x, y, z = sp.symbols(
    "a, m, p, q, r, s, t, u, v, x, y, z", real=True
)
```

## Ejercicio 76
$$\frac{\frac{1}{x^2} - x}{\frac{1}{x^2} + x}$$

```python
expr = ((1 / x**2) - x) / ((1 / x**2) + x)
display(sp.Eq(expr, expr.simplify()))
```

## Ejercicio 77
$$\frac{m+\frac{1}{2}}{2+\frac{1}{m}}$$

```python
expr = (m + sp.Rational(1, 2)) / (2 + (1 / m))
display(sp.Eq(expr, expr.simplify()))
```

## Ejercicio 78
$$\frac{\frac{1}{p} + \frac{1}{q}}{\frac{1}{p} - \frac{1}{q}}$$

```python
expr = ((1 / p) + (1 / q)) / ((1 / p) - (1 / q))
display(sp.Eq(expr, expr.simplify()))
```

## Ejercicio 79
$$\frac{\frac{1}{u^2} - \frac{1}{v^2}}{u^2 v^2}$$

```python
expr = ((1 / u**2) - (1 / v**2)) / (u**2 * v**2)
display(sp.Eq(expr, expr.simplify()))
```

## Ejercicio 80
$$\frac{1 + \frac{1}{\sqrt{x}}}{1 + \frac{1}{\sqrt{y}}}$$

```python
expr = (1 + (1 / sp.sqrt(x))) / (1 + (1 / sp.sqrt(y)))
display(sp.Eq(expr, expr.simplify()))
```

## Ejercicio 81

## Ejercicio 83
$$\frac{\frac{x}{x+4}}{\frac{x+5}{x}}$$

```python
expr_raw = r"\frac{\frac{x}{x+4}}{\frac{x+5}{x}}"
expr_sp = (x / (x + 4)) / ((x + 5) / x)
sol = expr_sp.expand()

print("raw expr:   \tSymPy expr:  \t\tsolution:")
print("-" * 55)

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad"  # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```

## Ejercicio 84
$$\frac{\frac{x^2+2x+1}{x}}{\frac{x^2 - 1}{x^2}}$$

```python
expr_raw = r"\frac{\frac{x^2+2x+1}{x}}{\frac{x^2 - 1}{x^2}}"
expr_sp = ((x**2 + 2 * x + 1) / x) / ((x**2 - 1) / x**2)
sol = expr_sp.factor()

print("raw expr:   \tSymPy expr:  \t\tsolution:")
print("-" * 55)

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad"  # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```

## Ejercicio 85
$$\frac{\frac{x^3 - 1}{x+1}}{\frac{x-1}{x^2 +2x +1}}$$

```python
expr_raw = r"\frac{\frac{x^3 - 1}{x+1}}{\frac{x-1}{x^2 +2x +1}}"
expr_sp = ((x**3 - 1) / (x + 1)) / ((x - 1) / (x**2 + 2 * x + 1))
sol = expr_sp.factor()

print("raw expr:   \tSymPy expr:  \t\tsolution:")
print("-" * 55)

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad"  # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```

## Ejercicio 90
$$\frac{\frac{a}{a-1} - \frac{a+1}{a}}{1- \frac{a}{a-1}}$$

```python
expr_raw = r"\frac{\frac{a}{a-1} - \frac{a+1}{a}}{1- \frac{a}{a-1}}"
expr_sp = (a / (a - 1) - ((a + 1) / a)) / (1 - (a / (a - 1)))
sol = expr_sp.simplify()

print("raw expr:   \tSymPy expr:  \t\tsolution:")
print("-" * 55)

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad"  # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```
