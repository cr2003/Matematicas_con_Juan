# Video 15 - Simplificar
$$\frac{\sqrt{7ab^2}}{\sqrt{49a}\sqrt{7b^4}}

```python
a, b = sp.symbols("a, b", positive=True) # Juan lo dice en el video para evitar numeros complejos
expr_raw = r"\frac{\sqrt{7ab^2}}{\sqrt{49a}\,\sqrt{7b^4}}"
expr_sp =(sp.sqrt(7*a*b**2)) / (sp.sqrt(49*a) * sp.sqrt(7*b**4))
sol = expr_sp.simplify()
print("raw expr:   \tSymPy expr:  \tsolution:")
print("-" * 40)

# Nice display
spacing = r"\quad = \qquad" # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```
