# Video 8 - Fraccion expresada como suma de dos fracciones

$$\frac{2x+6}{x^2-1} = \frac{A}{x+1} + \frac{B}{x-1}$$

```python
A, B = sp.symbols('A B')

# 1. Definimos la igualdad de los numeradores
# 2x + 6 = A(x - 1) + B(x + 1)
lado_izquierdo = 2*x + 6
lado_derecho = A*(x - 1) + B*(x + 1)

# 2. Creamos la ecuación de la diferencia (debe ser cero para que sean iguales)
# Esto nos permite extraer los coeficientes de x
ecuacion_polinomial = lado_izquierdo - lado_derecho

# 3. Generamos un sistema de ecuaciones igualando coeficientes de cada potencia de x a cero
# sp.Poly(..., x).coeffs() extrae los coeficientes
sistema = sp.Poly(ecuacion_polinomial, x).coeffs()

# 4. Resolvemos el sistema para A y B
soluciones = sp.solve(sistema, (A, B))

# Mostrar resultados
display(Math(rf"A = {sp.latex(soluciones[A])}, \quad B = {sp.latex(soluciones[B])}"))

# Verificación con apart()
fraccion = (2*x + 6) / (x**2 - 1)
# display(Math(rf"\text{{Comprobación con apart: }} {sp.latex(sp.apart(fraccion))}"))
display(Math(rf"\text{{Comprobación con apart: }} {sp.latex(fraccion)} = {sp.latex(sp.apart(fraccion))}"))
```
