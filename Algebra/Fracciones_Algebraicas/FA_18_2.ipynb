## Ejercicio 64
$$\frac{x^3 -8}{x^2 +3x -10}$$

```python
import sympy as sp
from IPython.display import display, Math

x = sp.symbols('x')

# 1. Corregimos el enunciado según la pizarra de Juan:
# El numerador es x^3 - 8 (que es 2 al cubo)
# El denominador es x^2 + 3x - 10
expr_raw = r"\frac{x^3 - 8}{x^2 + 3x - 10}"
expr_sp = (x**3 - 8) / (x**2 + 3*x - 10)

# 2. Simplificamos
sol = sp.cancel(expr_sp)

print("raw expr:   \tSymPy expr:  \tsolution:")
print("-" * 40)
spacing = r"\quad = \qquad" # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```
