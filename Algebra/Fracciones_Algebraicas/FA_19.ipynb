```python
from vpython import *
from astropy import units as u
import sympy as sp
from IPython.display import display, Math

a, m, n, p, q, r, s, t, u, v, x, y, z = sp.symbols(
    "a, m, n, p, q, r, s, t, u, v, x, y, z", real=True
)
```

## Ejercicio 1
$$\frac{x^2 +2x +1}{x^2 -1}$$


```python
expr = (x**2 + 2 * x + 1) / (x**2 - 1)
display(sp.Eq(expr, expr.factor()))
```

## Ejercicio 2
$$\frac{x^2 -2x +1}{x^2 -1}$$



```python
expr = (x**2 - 2 * x + 1) / (x**2 - 1)
display(sp.Eq(expr, expr.factor()))
```

## Ejercicio 3
$$\frac{4x^2 +4x +1}{2x +1}$$



```python
expr = (4 * x**2 + 4 * x + 1) / (2 * x + 1)
display(sp.Eq(expr, expr.factor()))
```

## Ejercicio 4
$$\frac{9x^2 -4}{3x -2}$$


```python
expr = (9 * x**2 - 4) / (3 * x - 2)
display(sp.Eq(expr, expr.factor()))
```

## Ejercicio 5
$$\frac{x^3 -1}{x -1}$$


```python
expr = (x**3 - 1) / (x - 1)
display(sp.Eq(expr, expr.factor()))
```

## Ejercicio 6
$$\frac{x^2 +x +1}{x^3 -1}$$


```python
expr = (x**2 + x + 1) / (x**3 - 1)
display(sp.Eq(expr, expr.factor()))
```
