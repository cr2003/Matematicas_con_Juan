# Video 2 - Simplificar Famosa Expresión

En óptica, la siguiente expresión se llama distancia focal de una lente:

$$\frac{1}{\frac{1}{p} + \frac{1}{q}}$$

**p**: distancia del objeto a la lente

**q**: distancia de la imagen a la lente

```python
expr = (1) / ((1/p) + (1/q))
display(sp.Eq(expr, expr.simplify()))
```
