# Video 11 - Simplificar
$$\frac{x^2 +y^2 -z^2 +2xy}{x^2 -y^2 +z^2 +2xz}$$

```python
expr = (x**2 + y**2 -z**2 +2*x*y) / (x**2 - y**2 +z**2 +2*x*z)
display(sp.Eq(expr, expr.simplify()))
```

### Tarea
$$\frac{1 -y^2}{(1+xy)^2 - (x+y)^2}$$

```python
expr = (1 - y**2) / ((1 + x*y)**2 - (x + y)**2)
display(sp.Eq(expr, expr.simplify()))
```
