# Video 10 - Libro escolar ruso II
$$\frac{(2x + 3y)^2 -4x^2 - 9y^2}{6xy}$$

```python
expr = ((2*x + 3*y)**2 - 4*x**2 -9*y**2) / (6*x*y)
display(sp.Eq(expr, expr.simplify()))
```

$$\frac{9x^2 +4y^2 -(3x-2y)^2}{-4xy}$$

```python
expr = (9*x**2 +4*y**2 -(3*x -2*y)**2) / (-4*x*y)
display(sp.Eq(expr, expr.simplify()))
```
