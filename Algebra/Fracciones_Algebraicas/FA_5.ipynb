# Video 5 - Simplifica

$$\frac{a-x}{a+x} + \frac{a+x}{a-x}$$

```python
expr = ((a-x) / (a+x)) + ((a+x) / (a - x))
sol = sp.simplify(sp.cancel(expr))
display(sp.Eq(expr, sol))
```
