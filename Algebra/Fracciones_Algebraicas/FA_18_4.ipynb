```python
from vpython import *
from astropy import units as u
import sympy as sp
from IPython.display import display, Math

a, m, n, p, q, r, s, t, u, v, x, y, z = sp.symbols(
    "a, m, n, p, q, r, s, t, u, v, x, y, z", real=True
)
```

## Ejercicio 91
$$\frac{5m^3}{15m^4}$$


```python
expr_raw = r"\frac{5m^3}{15m^4}"
expr_sp = (5 * m**3) / (15 * m**4)
sol = expr_sp.simplify()

print("raw expr:   \tSymPy expr:  \t\tsolution:")
print("-" * 55)

# Sustituye tu línea 7 por esta:
spacing = r"\quad = \qquad"  # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```

## Ejercicio 92
$$\frac{7p^3q^3 -4p^2q^5}{p^2 q^3}$$


```python
expr = (7 * p**3 * q**3 - 4 * p**2 * q**5) / (p**2 * q**3)
display(sp.Eq(expr, expr.factor()))
```

## Ejercicio 93
$$\frac{m^2 -2}{m + \sqrt{2}}$$


```python
expr_raw = r"\frac{m^2 -2}{m + \sqrt{2}}"
expr_sp = (m**2 - 2) / (m + sp.sqrt(2))
# FORZAMOS la factorización extendiendo el dominio a la raíz de 2
# Esto permite que SymPy factorice (m² - 2) como (m - √2)(m + √2)
sol = sp.factor(expr_sp, extension=[sp.sqrt(2)])

print("raw expr:   \tSymPy expr:  \t\tsolution:")
print("-" * 55)

spacing = r"\quad = \qquad"  # Ajusta la cantidad de \qquad según necesites
display(Math(rf"{expr_raw} {spacing} {sp.latex(expr_sp)} {spacing} {sp.latex(sol)}"))
```

## Ejercicio 99
$$\frac{\frac{1}{m^2} - \frac{1}{n^2}}{n - m}$$
