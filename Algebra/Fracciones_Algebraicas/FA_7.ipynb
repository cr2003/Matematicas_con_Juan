# Video 7 - Simplificar
$$\frac{\sqrt{x}}{2+\sqrt{x}} 

```python
expr = (sp.sqrt(x)) / (2 + sp.sqrt(x))
# radsimp es la función específica para racionalizar denominadores
sol_raw = sp.radsimp(expr)
num, den = sp.fraction(sol_raw)
# Cambiamos el signo a ambos manualmente
sol_estetica = (-num) / (-den)

display(sp.Eq(expr, sol_estetica, evaluate=False))
```
